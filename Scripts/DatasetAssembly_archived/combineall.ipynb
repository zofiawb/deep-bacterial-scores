```python
import pandas as pd
from pathlib import Path

# Directory containing CSVs
indir = Path("combined_datasets")
outfile = Path("combined_datasets/all_species_combined.csv")

# Gather all CSV files in that folder
files = sorted(indir.glob("*.csv"))
print(f"Found {len(files)} CSVs:", [f.name for f in files])

dfs = []
for f in files:
    try:
        df = pd.read_csv(f, low_memory=False)
    except Exception as e:
        print(f"[ERROR] reading {f.name}: {e}")
        continue

    # Use the file stem (name without .csv) as a prefix
    prefix = f.stem
    df = df.add_prefix(prefix + "_")

    dfs.append(df)

combined = pd.concat(dfs, axis=0, ignore_index=True)

combined.to_csv(outfile, index=False)
print(f"Wrote combined CSV → {outfile}")
```
